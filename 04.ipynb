```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Membaca file data
df = pd.read_csv('Kelas G_Fixed Call Report (1).csv')

# 2. Mengatur ukuran grafik dan tema visual
plt.figure(figsize=(10, 6))
sns.set_theme(style="whitegrid")

# 3. Memilih kolom yang ingin dibuatkan boxplot
df_boxplot = df[['Number of Total Complaints', 'Number of Calls To Know About The Service']]

# 4. Membuat Boxplot menggunakan seaborn
sns.boxplot(data=df_boxplot, palette="Set3", width=0.5)

# 5. Menambahkan Judul dan Label Sumbu
plt.title('Boxplot Distribusi Data Komplain vs Panggilan Informasi Layanan', fontsize=14, fontweight='bold', pad=15)
plt.ylabel('Jumlah Kasus / Panggilan', fontsize=12)
plt.xticks(fontsize=11)

# 6. Menampilkan Grafik
plt.tight_layout()
plt.show()
```
