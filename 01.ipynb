```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Membaca file data
df = pd.read_csv('Kelas G_Fixed Call Report (1).csv')

# 2. Mengelompokkan dan menjumlahkan data berdasarkan Tahun
df_yearly = df.groupby('Year')[['Number of Total Complaints', 'Number of Calls To Know About The Service']].sum().reset_index()

# 3. Mengubah format data menjadi 'long-format' agar sesuai dengan library seaborn
df_melted = df_yearly.melt(
    id_vars='Year', 
    value_vars=['Number of Total Complaints', 'Number of Calls To Know About The Service'],
    var_name='Jenis Panggilan', 
    value_name='Total Jumlah'
)

# 4. Mengatur tema dan ukuran grafik
plt.figure(figsize=(10, 6))
sns.set_theme(style="whitegrid")

# 5. Membuat Diagram Batang (Bar Chart)
ax = sns.barplot(
    data=df_melted, 
    x='Year', 
    y='Total Jumlah', 
    hue='Jenis Panggilan', 
    palette='Set2'
)

# 6. Menambahkan angka di atas setiap batang (opsional, agar lebih informatif)
for container in ax.containers:
    ax.bar_label(container, fmt='%d', padding=3, fontsize=9)

# 7. Menambahkan Judul dan Label Sumbu
plt.title('Perbandingan Total Komplain vs Panggilan Informasi Layanan per Tahun', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Tahun', fontsize=12)
plt.ylabel('Total Akumulasi', fontsize=12)
plt.legend(title='Kategori')

# 8. Menampilkan Grafik
plt.tight_layout()
plt.show()
```
